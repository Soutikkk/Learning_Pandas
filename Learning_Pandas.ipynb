```python
import pandas as pd

data = {
    "Name": ["SpongeBob", "Patrick", "Sandy", "Squidward", "Mr. Krabs"],
    "Age": [20, 21, 19, 30, 45],
    "Occupation": ["Fry Cook", "Unemployed", "Scientist", "Cashier", "Business Owner"]
}

dataframe = pd.DataFrame(
    data,
    index=["Employee 1", "Employee 2", "Employee 3", "Employee 4", "Employee 5"]
)

# Add a new column
dataframe["Salary"] = [30000, 25000, 40000, 35000, 50000]

# Add a new row
new_employee = pd.DataFrame(
    {
        "Name": ["Plankton"],
        "Age": [35],
        "Occupation": ["Restaurant Owner"],
        "Salary": [60000]
    },
    index=["Employee 6"]
)

dataframe = pd.concat([dataframe, new_employee])

print(dataframe)
```

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv", index_col="Name")

water_pokemon = df[df["Type1"] == "Water"]

print(water_pokemon)    
```

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv", index_col="Name")

water_pokemon = df[(df["Type1"] == "Water") | (df["Type2"] == "Water")]

print(water_pokemon)    
```

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv", index_col="Name")

fire_pokemon = df[(df["Type1"] == "Fire") & (df["Type2"] == "Flying")]

print(fire_pokemon)    
```

Data Cleaning = The process of fixing/removing,
incomplete ,incorrect, or irrelevant data.
~75% of work done with Pandas is data Cleaning

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv")

df = df.drop(columns=["Legendary"])

print(df)
```

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv")

df = df.drop(columns=["Legendary", "No"], errors="ignore")

print(df)
```

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv")

df = df.dropna(subset=["Type2"])

print(df.to_string())
```

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv")

df = df.fillna({"Type2" : "None"})

print(df.to_string())
```

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv")

df["Type2"] = df["Type2"].replace({"Grass": "GRASS"})

print(df.to_string())
```

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv")

df["Type1"] = df["Type1"].replace({"Grass": "GRASS", "Fire": "FIRE", "Water": "WATER"})

print(df.to_string())
```

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv")

df["Name"] = df["Name"].str.upper()

print(df.to_string())
```

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv")

df["Name"] = df["Name"].str.upper()

print(df.to_string())
```

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv")

df["Name"] = df["Name"].str.lower()

print(df.to_string())
```

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv")

df["Legendary"] = df["Legendary"].astype(bool)

print(df.to_string())
```

```python
import pandas as pd

df = pd.read_csv("Pokemon.csv")

df = df.drop_duplicates()

print(df.to_string())
```
